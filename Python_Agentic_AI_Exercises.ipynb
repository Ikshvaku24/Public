### Solution 5: Agent with Memory

````python
# SOLUTION - Exercise 5
class MemoryAgent:
    def __init__(self, name, max_memory=5):
        self.name = name
        self.max_memory = max_memory
        self.memory = []
    
    def add_message(self, message):
        self.memory.append(message)
        # Remove oldest if exceeds limit
        if len(self.memory) > self.max_memory:
            self.memory.pop(0)
    
    def get_memory(self):
        return self.memory
    
    def clear_memory(self):
        self.memory = []
    
    def get_memory_summary(self):
        return {
            "agent_name": self.name,
            "messages_stored": len(self.memory),
            "memory_limit": self.max_memory
        }

# Test
agent = MemoryAgent("TestBot", max_memory=3)
agent.add_message("Message 1")
agent.add_message("Message 2")
agent.add_message("Message 3")
agent.add_message("Message 4")
agent.add_message("Message 5")

print("Memory:", agent.get_memory())  # Should show last 3
print("Summary:", agent.get_memory_summary())
````

### Solution 6: Tool Execution with Error Handling

````python
# SOLUTION - Exercise 6
class ToolAgent:
    def __init__(self, name):
        self.name = name
    
    def execute_tool(self, tool_name, **parameters):
        try:
            if tool_name == "calculator":
                result = parameters["a"] + parameters["b"]
                return {"status": "success", "result": result}
            
            elif tool_name == "multiply":
                result = parameters["a"] * parameters["b"]
                return {"status": "success", "result": result}
            
            elif tool_name == "power":
                result = parameters["base"] ** parameters["exponent"]
                return {"status": "success", "result": result}
            
            else:
                return {"status": "error", "message": f"Unknown tool: {tool_name}"}
        
        except KeyError as e:
            return {"status": "error", "message": f"Missing parameter: {e}"}
        except Exception as e:
            return {"status": "error", "message": f"Error: {str(e)}"}

# Test
agent = ToolAgent("ToolBot")
print(agent.execute_tool("calculator", a=5, b=3))      # Success
print(agent.execute_tool("multiply", a=4, b=7))        # Success
print(agent.execute_tool("power", base=2, exponent=8)) # Success
print(agent.execute_tool("calculator", a=5))           # Missing parameter
print(agent.execute_tool("unknown", a=1))              # Unknown tool
````

### Solution 7: Agent State Persistence

````python
# SOLUTION - Exercise 7
import json

class PersistentAgent:
    def __init__(self, name):
        self.name = name
        self.task_list = []
        self.task_count = 0
    
    def add_task(self, task):
        self.task_list.append(task)
        self.task_count += 1
    
    def save_state(self, filename):
        try:
            state = {
                "name": self.name,
                "task_count": self.task_count,
                "task_list": self.task_list
            }
            with open(filename, 'w') as f:
                json.dump(state, f, indent=2)
            return True
        except Exception as e:
            print(f"Error saving: {e}")
            return False
    
    def load_state(self, filename):
        try:
            with open(filename, 'r') as f:
                state = json.load(f)
            
            self.name = state["name"]
            self.task_count = state["task_count"]
            self.task_list = state["task_list"]
            return True
        
        except FileNotFoundError:
            print(f"File {filename} not found")
            return False
        except json.JSONDecodeError:
            print("Invalid JSON file")
            return False
        except Exception as e:
            print(f"Error loading: {e}")
            return False

# Test
agent1 = PersistentAgent("Agent1")
agent1.add_task("Task A")
agent1.add_task("Task B")
agent1.add_task("Task C")
print(f"Agent1: {agent1.name}, Tasks: {agent1.task_count}")

agent1.save_state("agent_state_test.json")

agent2 = PersistentAgent("Agent2")
print(f"\nAgent2 before load: {agent2.name}, Tasks: {agent2.task_count}")

agent2.load_state("agent_state_test.json")
print(f"Agent2 after load: {agent2.name}, Tasks: {agent2.task_count}")
print(f"Task list: {agent2.task_list}")
````

### Solution 9: Multi-Agent Coordinator

````python
# SOLUTION - Exercise 9
import json

class AgentCoordinator:
    def __init__(self):
        self.agents = {}
        self.task_log = []
    
    def register_agent(self, agent_id, agent_name, capabilities):
        self.agents[agent_id] = {
            "name": agent_name,
            "capabilities": capabilities,
            "tasks_completed": 0
        }
        print(f"✓ Registered agent: {agent_name} (ID: {agent_id})")
    
    def assign_task(self, agent_id, task_description, required_capability):
        try:
            # Check if agent exists
            if agent_id not in self.agents:
                raise KeyError(f"Agent {agent_id} not found")
            
            agent = self.agents[agent_id]
            
            # Check capability
            if required_capability not in agent["capabilities"]:
                raise ValueError(f"Agent {agent_id} doesn't have capability: {required_capability}")
            
            # Assign task
            agent["tasks_completed"] += 1
            self.task_log.append({
                "agent_id": agent_id,
                "task": task_description,
                "capability": required_capability,
                "status": "success"
            })
            return {"status": "success"}
        
        except (KeyError, ValueError) as e:
            self.task_log.append({
                "agent_id": agent_id,
                "task": task_description,
                "capability": required_capability,
                "status": "failed"
            })
            return {"status": "error", "message": str(e)}
    
    def get_agent_stats(self, agent_id):
        if agent_id in self.agents:
            return self.agents[agent_id]
        return {"error": "Agent not found"}
    
    def get_top_performer(self):
        if not self.agents:
            return None
        return max(self.agents.items(), key=lambda x: x[1]["tasks_completed"])[0]
    
    def save_coordinator_state(self, filename):
        state = {
            "agents": self.agents,
            "task_log": self.task_log
        }
        with open(filename, 'w') as f:
            json.dump(state, f, indent=2)
        print(f"✓ State saved to {filename}")

# Test
coordinator = AgentCoordinator()

# Register agents
coordinator.register_agent("agent_1", "Researcher", ["search", "analyze"])
coordinator.register_agent("agent_2", "Coder", ["code", "debug"])
coordinator.register_agent("agent_3", "Writer", ["write", "summarize"])

print("\n" + "="*50)
# Assign tasks
print(coordinator.assign_task("agent_1", "Research AI trends", "search"))
print(coordinator.assign_task("agent_1", "Analyze data", "analyze"))
print(coordinator.assign_task("agent_2", "Write Python script", "code"))
print(coordinator.assign_task("agent_2", "Fix bugs", "debug"))
print(coordinator.assign_task("agent_2", "Debug more", "debug"))
print(coordinator.assign_task("agent_3", "Write report", "write"))
print(coordinator.assign_task("agent_1", "Write code", "code"))  # Should fail
print(coordinator.assign_task("agent_99", "Task", "any"))       # Should fail

print("\n" + "="*50)
print(f"Top performer: {coordinator.get_top_performer()}")
print(f"Agent 2 stats: {coordinator.get_agent_stats('agent_2')}")

coordinator.save_coordinator_state("coordinator_state.json")
````

### Solution 10: Conversation Agent with Context

````python
# SOLUTION - Exercise 10
import json

class ContextualAgent:
    def __init__(self, name, personality="helpful"):
        self.name = name
        self.personality = personality
        self.conversation_history = []
        self.user_context = {}
        self.timestamp_counter = 0
    
    def receive_message(self, message):
        # Add user message
        self.timestamp_counter += 1
        self.conversation_history.append({
            "role": "user",
            "content": message,
            "timestamp": self.timestamp_counter
        })
        
        # Extract context
        message_lower = message.lower()
        
        if "my name is" in message_lower:
            parts = message_lower.split("my name is")
            if len(parts) > 1:
                name = parts[1].strip().split()[0].capitalize()
                self.user_context["name"] = name
        
        if "i like" in message_lower:
            parts = message_lower.split("i like")
            if len(parts) > 1:
                interest = parts[1].strip()
                if "interests" not in self.user_context:
                    self.user_context["interests"] = []
                self.user_context["interests"].append(interest)
        
        if "i am from" in message_lower:
            parts = message_lower.split("i am from")
            if len(parts) > 1:
                location = parts[1].strip()
                self.user_context["location"] = location
        
        # Generate response
        response = self._generate_response(message)
        
        # Add assistant message
        self.timestamp_counter += 1
        self.conversation_history.append({
            "role": "assistant",
            "content": response,
            "timestamp": self.timestamp_counter
        })
        
        return response
    
    def _generate_response(self, message):
        message_lower = message.lower()
        
        if "my name is" in message_lower and "name" in self.user_context:
            return f"Nice to meet you, {self.user_context['name']}! I'm {self.name}."
        
        if "what do you know about me" in message_lower:
            if self.user_context:
                info = []
                if "name" in self.user_context:
                    info.append(f"Your name is {self.user_context['name']}")
                if "interests" in self.user_context:
                    info.append(f"You like {', '.join(self.user_context['interests'])}")
                if "location" in self.user_context:
                    info.append(f"You're from {self.user_context['location']}")
                return ". ".join(info) + "."
            return "I don't know much about you yet. Tell me more!"
        
        if "i like" in message_lower:
            return "That's interesting! Thanks for sharing."
        
        if "i am from" in message_lower:
            return f"Nice! {self.user_context.get('location', 'Your location')} sounds great!"
        
        return f"I understand. How can I help you further?"
    
    def get_conversation_stats(self):
        total = len(self.conversation_history)
        user_count = len([m for m in self.conversation_history if m["role"] == "user"])
        assistant_count = total - user_count
        
        if total > 0:
            avg_length = sum(len(m["content"]) for m in self.conversation_history) / total
        else:
            avg_length = 0
        
        return {
            "total_messages": total,
            "user_messages": user_count,
            "assistant_messages": assistant_count,
            "avg_message_length": round(avg_length, 2)
        }
    
    def search_conversation(self, keyword):
        return [msg for msg in self.conversation_history 
                if keyword.lower() in msg["content"].lower()]
    
    def export_conversation(self, filename):
        data = {
            "conversation_history": self.conversation_history,
            "user_context": self.user_context
        }
        with open(filename, 'w') as f:
            json.dump(data, f, indent=2)
        print(f"✓ Exported to {filename}")

# Test
agent = ContextualAgent("Alex", personality="friendly")
print(agent.receive_message("Hi, my name is Rahul"))
print(agent.receive_message("I like AI and programming"))
print(agent.receive_message("I am from Delhi"))
print(agent.receive_message("What do you know about me?"))

print("\n" + "="*50)
print("Stats:", agent.get_conversation_stats())
print("\nSearch 'AI':", agent.search_conversation("AI"))
print("\nUser Context:", agent.user_context)

agent.export_conversation("conversation_export.json")
````

### Solution 11: Robust Tool Agent with Validation

````python
# SOLUTION - Exercise 11
class SmartToolAgent:
    def __init__(self, name):
        self.name = name
        self.tool_registry = {}
        self.execution_history = []
        self.execution_counter = 0
        
        # Register default tools
        self._register_default_tools()
    
    def _register_default_tools(self):
        # Calculator tool
        def calc_validator(params):
            if params.get('operation') not in ['add', 'subtract', 'multiply', 'divide']:
                raise ValueError("Invalid operation")
            if params.get('operation') == 'divide' and params.get('b') == 0:
                raise ValueError("Cannot divide by zero")
        
        self.register_tool("calculator", ["a", "b", "operation"], calc_validator)
        
        # Text processor
        def text_validator(params):
            if params.get('operation') not in ['count_words', 'reverse', 'uppercase']:
                raise ValueError("Invalid text operation")
        
        self.register_tool("text_processor", ["text", "operation"], text_validator)
        
        # Data analyzer
        def data_validator(params):
            if not isinstance(params.get('numbers'), list):
                raise ValueError("Numbers must be a list")
            if len(params.get('numbers', [])) == 0:
                raise ValueError("Numbers list cannot be empty")
        
        self.register_tool("data_analyzer", ["numbers", "operation"], data_validator)
    
    def register_tool(self, tool_name, required_params, validator_func=None):
        self.tool_registry[tool_name] = {
            "required_params": required_params,
            "validator": validator_func,
            "call_count": 0,
            "success_count": 0
        }
    
    def execute_tool(self, tool_name, **params):
        self.execution_counter += 1
        execution_id = self.execution_counter
        
        try:
            # Check tool exists
            if tool_name not in self.tool_registry:
                raise ValueError(f"Unknown tool: {tool_name}")
            
            tool = self.tool_registry[tool_name]
            tool["call_count"] += 1
            
            # Check required parameters
            missing = [p for p in tool["required_params"] if p not in params]
            if missing:
                raise KeyError(f"Missing parameters: {missing}")
            
            # Run validator
            if tool["validator"]:
                tool["validator"](params)
            
            # Execute tool logic
            result = self._execute_tool_logic(tool_name, **params)
            
            # Log success
            tool["success_count"] += 1
            self.execution_history.append({
                "id": execution_id,
                "tool": tool_name,
                "params": params,
                "status": "success",
                "result": result
            })
            
            return {"status": "success", "result": result}
        
        except Exception as e:
            # Log failure
            self.execution_history.append({
                "id": execution_id,
                "tool": tool_name,
                "params": params,
                "status": "failed",
                "error": str(e)
            })
            
            return {"status": "error", "message": str(e)}
    
    def _execute_tool_logic(self, tool_name, **params):
        if tool_name == "calculator":
            a, b = params['a'], params['b']
            op = params['operation']
            if op == 'add':
                return a + b
            elif op == 'subtract':
                return a - b
            elif op == 'multiply':
                return a * b
            elif op == 'divide':
                return a / b
        
        elif tool_name == "text_processor":
            text = params['text']
            op = params['operation']
            if op == 'count_words':
                return len(text.split())
            elif op == 'reverse':
                return text[::-1]
            elif op == 'uppercase':
                return text.upper()
        
        elif tool_name == "data_analyzer":
            numbers = params['numbers']
            op = params['operation']
            if op == 'max':
                return max(numbers)
            elif op == 'min':
                return min(numbers)
            elif op == 'average':
                return sum(numbers) / len(numbers)
    
    def get_tool_stats(self):
        stats = {}
        for tool_name, tool_data in self.tool_registry.items():
            success_rate = 0
            if tool_data["call_count"] > 0:
                success_rate = (tool_data["success_count"] / tool_data["call_count"]) * 100
            
            stats[tool_name] = {
                "call_count": tool_data["call_count"],
                "success_count": tool_data["success_count"],
                "success_rate": round(success_rate, 2)
            }
        return stats
    
    def get_failed_executions(self):
        return [ex for ex in self.execution_history if ex["status"] == "failed"]

# Test
agent = SmartToolAgent("ToolMaster")

print("=== Testing Calculator ===")
print(agent.execute_tool("calculator", a=10, b=5, operation="add"))
print(agent.execute_tool("calculator", a=10, b=5, operation="multiply"))
print(agent.execute_tool("calculator", a=10, b=0, operation="divide"))  # Error
print(agent.execute_tool("calculator", a=10, operation="add"))         # Missing param

print("\n=== Testing Text Processor ===")
print(agent.execute_tool("text_processor", text="Hello World", operation="count_words"))
print(agent.execute_tool("text_processor", text="Python", operation="reverse"))
print(agent.execute_tool("text_processor", text="ai agent", operation="uppercase"))

print("\n=== Testing Data Analyzer ===")
print(agent.execute_tool("data_analyzer", numbers=[1, 5, 3, 9, 2], operation="max"))
print(agent.execute_tool("data_analyzer", numbers=[10, 20, 30], operation="average"))
print(agent.execute_tool("data_analyzer", numbers=[], operation="max"))  # Error

print("\n=== Tool Statistics ===")
import json
print(json.dumps(agent.get_tool_stats(), indent=2))

print("\n=== Failed Executions ===")
for failure in agent.get_failed_executions():
    print(f"ID {failure['id']}: {failure['tool']} - {failure['error']}")
````
