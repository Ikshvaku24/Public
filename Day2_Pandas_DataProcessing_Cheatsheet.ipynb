## 📌 Selecting Data

### Column Selection

```python
df['col']               # Single column (Series)
df[['col']]             # Single column (DataFrame)
df[['col1', 'col2']]    # Multiple columns
df.col                  # Attribute access (Series)
```

### Row Selection

```python
# iloc - integer position
df.iloc[0]              # First row
df.iloc[-1]             # Last row
df.iloc[0:5]            # First 5 rows
df.iloc[[0, 2, 4]]      # Specific rows
df.iloc[0:3, 0:2]       # Rows 0-2, columns 0-1

# loc - label-based
df.loc[0]               # Row with index 0
df.loc[0:2]             # Rows 0 to 2 (inclusive)
df.loc[[0, 2, 4]]       # Specific index labels
df.loc[0:2, ['col1', 'col2']]  # Rows and columns
df.loc[df['A'] > 5]     # Conditional selection
```

## 📌 Data Cleaning

### Handling Missing Values

```python
# Detecting nulls
df.isnull()             # Boolean DataFrame
df.notnull()            # Opposite
df.isnull().sum()       # Count per column
df.isnull().any()       # Any null per column?

# Dropping nulls
df.dropna()             # Drop rows with any null
df.dropna(axis=1)       # Drop columns with any null
df.dropna(how='all')    # Drop if all values are null
df.dropna(subset=['col1', 'col2'])  # Drop if null in specific columns
df.dropna(thresh=3)     # Keep rows with at least 3 non-null values

# Filling nulls
df.fillna(0)            # Fill all nulls with 0
df.fillna({'A': 0, 'B': 1})  # Different values per column
df.fillna(method='ffill')    # Forward fill
df.fillna(method='bfill')    # Backward fill
df['col'].fillna(df['col'].mean())  # Fill with mean
df['col'].fillna(df['col'].median())  # Fill with median
df['col'].fillna(df['col'].mode()[0])  # Fill with mode
```

### Handling Duplicates

```python
# Detecting duplicates
df.duplicated()         # Boolean Series
df.duplicated().sum()   # Count duplicates
df[df.duplicated()]     # Show duplicate rows
df[df.duplicated(keep=False)]  # Show all including first

# Removing duplicates
df.drop_duplicates()    # Keep first occurrence
df.drop_duplicates(keep='last')  # Keep last
df.drop_duplicates(keep=False)   # Remove all duplicates
df.drop_duplicates(subset=['col1'])  # Based on specific columns
```

### Type Conversions

```python
# Changing types
df['col'].astype(int)           # To integer
df['col'].astype(float)         # To float
df['col'].astype(str)           # To string
df['col'].astype('category')    # To categorical

# Safe conversion
pd.to_numeric(df['col'], errors='coerce')  # Invalid → NaN
pd.to_datetime(df['col'], errors='coerce')  # Invalid → NaT

# DateTime conversion
pd.to_datetime(df['col'])
pd.to_datetime(df['col'], format='%Y-%m-%d')
pd.to_datetime(df['col'], dayfirst=True)
```

## 📌 GroupBy Operations

```python
# Basic groupby
df.groupby('col').sum()                    # Sum by group
df.groupby('col')['value'].sum()           # Single column
df.groupby(['col1', 'col2']).sum()         # Multiple groups

# Aggregation functions
df.groupby('col').agg('sum')               # Sum
df.groupby('col').agg(['sum', 'mean', 'count'])
df.groupby('col').agg({
    'col1': 'sum',
    'col2': ['mean', 'max']
})

# Custom aggregations
df.groupby('col').agg(
    total=('value', 'sum'),
    avg=('value', 'mean'),
    custom=('value', lambda x: x.max() - x.min())
)

# Transform (returns same-sized DataFrame)
df.groupby('col')['value'].transform('mean')  # Group mean for each row
df['pct_of_group'] = df['value'] / df.groupby('col')['value'].transform('sum')

# Filter groups
df.groupby('col').filter(lambda x: len(x) > 5)  # Groups with >5 members

# Common aggregations
.sum()      .mean()     .median()   .std()
.min()      .max()      .count()    .size()
.first()    .last()     .nunique()  .var()
```

## 📌 Merging & Joining

### Merge

```python
# Basic merge
pd.merge(df1, df2, on='key')                    # Inner join (default)
pd.merge(df1, df2, on='key', how='left')        # Left join
pd.merge(df1, df2, on='key', how='right')       # Right join
pd.merge(df1, df2, on='key', how='outer')       # Outer join

# Multiple keys
pd.merge(df1, df2, on=['key1', 'key2'])

# Different column names
pd.merge(df1, df2, left_on='key1', right_on='key2')

# Merge on index
pd.merge(df1, df2, left_index=True, right_index=True)

# Suffix for duplicate columns
pd.merge(df1, df2, on='key', suffixes=('_left', '_right'))

# Indicator column (shows merge source)
pd.merge(df1, df2, on='key', how='outer', indicator=True)
```

### Concat

```python
# Vertical (stack rows)
pd.concat([df1, df2])                      # Default axis=0
pd.concat([df1, df2], ignore_index=True)   # Reset index

# Horizontal (side by side)
pd.concat([df1, df2], axis=1)

# Only keep common columns/rows
pd.concat([df1, df2], join='inner')

# Add keys to identify source
pd.concat([df1, df2], keys=['df1', 'df2'])
```

### Join

```python
# Join on index
df1.join(df2)                    # Left join (default)
df1.join(df2, how='inner')       # Inner join
df1.join(df2, lsuffix='_l', rsuffix='_r')  # Suffix for duplicates
```

## 📌 Pivot Tables & Reshaping

### Pivot Table

```python
# Basic pivot
df.pivot_table(
    values='value_col',
    index='row_col',
    columns='col_col',
    aggfunc='sum'        # sum, mean, count, etc.
)

# Multiple aggregations
df.pivot_table(
    values='value_col',
    index='row_col',
    columns='col_col',
    aggfunc=['sum', 'mean', 'count']
)

# Multiple values
df.pivot_table(
    values=['val1', 'val2'],
    index='row_col',
    columns='col_col',
    aggfunc='sum'
)

# With totals
df.pivot_table(
    values='value_col',
    index='row_col',
    columns='col_col',
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)

# Fill missing values
df.pivot_table(..., fill_value=0)
```

### Cross-Tabulation

```python
# Frequency table
pd.crosstab(df['row_col'], df['col_col'])

# With values
pd.crosstab(
    df['row_col'],
    df['col_col'],
    values=df['value_col'],
    aggfunc='sum'
)

# With percentages
pd.crosstab(df['row'], df['col'], normalize='all')     # All cells
pd.crosstab(df['row'], df['col'], normalize='index')   # Row-wise
pd.crosstab(df['row'], df['col'], normalize='columns') # Column-wise

# With margins
pd.crosstab(df['row'], df['col'], margins=True)
```

## 📌 Quick Reference: Common Tasks

### Load → Clean → Transform → Save

```python
# 1. LOAD
df = pd.read_csv('data.csv')

# 2. CLEAN
df = df.drop_duplicates()                       # Remove duplicates
df = df.dropna(subset=['important_col'])        # Drop nulls in key columns
df['col'] = df['col'].fillna(df['col'].mean())  # Fill remaining nulls

# 3. TRANSFORM
df['date'] = pd.to_datetime(df['date'])         # Convert types
df['new_col'] = df['A'] + df['B']               # Calculate new columns
df = df[df['value'] > 0]                        # Filter
df = df.sort_values('date')                     # Sort

# 4. AGGREGATE
summary = df.groupby('category').agg({
    'value': ['sum', 'mean', 'count']
})

# 5. SAVE
df.to_csv('output.csv', index=False)
df.to_parquet('output.parquet')
summary.to_excel('summary.xlsx')
```

## 📌 Advanced: Method Chaining

```python
# Clean, transform, and analyze in one chain
result = (
    pd.read_csv('data.csv')
    .drop_duplicates()
    .dropna(subset=['key_col'])
    .assign(
        date=lambda x: pd.to_datetime(x['date']),
        total=lambda x: x['quantity'] * x['price']
    )
    .query('total > 100')
    .groupby('category')['total']
    .sum()
    .sort_values(ascending=False)
)
```

**Remember:**
- Use `df.head()` and `df.info()` frequently to inspect your data

- Use appropriate file formats: Parquet for large data, Excel for reports
- Profile your code for performance bottlenecks
