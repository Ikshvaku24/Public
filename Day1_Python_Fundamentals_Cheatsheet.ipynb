## 📌 Data Structures

### Lists - Ordered, Mutable `[]`

```python
# Creation
my_list = [1, 2, 3, 4, 5]
empty_list = []

# Accessing
my_list[0]      # First element (index 0)
my_list[-1]     # Last element
my_list[1:4]    # Slicing [start:end] (end exclusive)

# Common methods
my_list.append(6)           # Add to end
my_list.insert(0, 0)        # Insert at position
my_list.remove(3)           # Remove first occurrence
my_list.pop()               # Remove and return last
my_list.pop(0)              # Remove and return at index
my_list.extend([7, 8])      # Add multiple elements
my_list.clear()             # Remove all elements
my_list.sort()              # Sort in place
my_list.reverse()           # Reverse in place
len(my_list)                # Length
3 in my_list                # Check membership
my_list.count(2)            # Count occurrences
my_list.index(3)            # Find index of element
```

### Tuples - Ordered, Immutable `()`

```python
# Creation
my_tuple = (1, 2, 3)
single_tuple = (1,)         # Single element needs comma

# Accessing (same as lists)
my_tuple[0]
my_tuple[-1]

# Unpacking
x, y, z = my_tuple

# Methods
my_tuple.count(2)           # Count occurrences
my_tuple.index(3)           # Find index
```

### Dictionaries - Key-Value Pairs `{}`

```python
# Creation
my_dict = {"name": "Alice", "age": 25, "city": "Delhi"}
empty_dict = {}

# Accessing
my_dict["name"]                     # Direct access (error if key missing)
my_dict.get("name")                 # Safe access (returns None if missing)
my_dict.get("salary", 0)            # With default value

# Adding/Modifying
my_dict["email"] = "a@b.com"        # Add new key
my_dict["age"] = 26                 # Modify existing

# Methods
my_dict.keys()                      # All keys
my_dict.values()                    # All values
my_dict.items()                     # Key-value pairs
my_dict.pop("age")                  # Remove and return value
my_dict.update({"role": "DE"})      # Merge dictionaries
"name" in my_dict                   # Check key existence
len(my_dict)                        # Number of keys
```

### Sets - Unordered, Unique Elements `{}`

```python
# Creation
my_set = {1, 2, 3, 4, 5}
empty_set = set()               # Use set(), not {}

# Methods
my_set.add(6)                   # Add element
my_set.remove(3)                # Remove (error if not exists)
my_set.discard(3)               # Remove (no error)
my_set.clear()                  # Remove all

# Set operations
set_a = {1, 2, 3}
set_b = {3, 4, 5}
set_a | set_b                   # Union: {1, 2, 3, 4, 5}
set_a & set_b                   # Intersection: {3}
set_a - set_b                   # Difference: {1, 2}
set_a ^ set_b                   # Symmetric diff: {1, 2, 4, 5}
```

## 📌 Comprehensions

```python
# List comprehension
[x**2 for x in range(10)]                       # Squares: [0, 1, 4, 9, ...]
[x for x in range(10) if x % 2 == 0]           # Even numbers
[x if x % 2 == 0 else 0 for x in range(10)]    # With condition

# Dictionary comprehension
{x: x**2 for x in range(5)}                     # {0: 0, 1: 1, 2: 4, ...}
{k: v for k, v in dict.items() if v > 10}      # Filter dict

# Set comprehension
{x**2 for x in range(10)}                       # Set of squares

# Generator expression (memory efficient)
(x**2 for x in range(1000000))                  # Lazy evaluation
```

## 📌 Exception Handling

```python
# Basic try-except
try:
    # Code that might raise exception
    risky_operation()
except ExceptionType:
    # Handle specific exception
    pass

# Multiple exceptions
try:
    operation()
except ValueError:
    # Handle ValueError
    pass
except TypeError:
    # Handle TypeError
    pass
except Exception as e:
    # Catch all other exceptions
    print(f"Error: {e}")

# Complete form
try:
    operation()
except ExceptionType as e:
    # Handle exception
    pass
else:
    # Runs if no exception
    pass
finally:
    # Always runs (cleanup)
    pass

# Raising exceptions
raise ValueError("Invalid value")

# Custom exception
class CustomError(Exception):
    pass
```

## 📌 File Operations

### Text Files

```python
# Reading
with open("file.txt", "r") as f:
    content = f.read()              # Read entire file
    lines = f.readlines()           # Read as list of lines
    line = f.readline()             # Read single line

# Writing
with open("file.txt", "w") as f:    # Overwrites file
    f.write("text\n")
    f.writelines(["line1\n", "line2\n"])

# Appending
with open("file.txt", "a") as f:    # Adds to end
    f.write("new line\n")

# File modes
# 'r'  - Read (default)
# 'w'  - Write (creates/overwrites)
# 'a'  - Append
# 'r+' - Read and write
# 'b'  - Binary mode (e.g., 'rb', 'wb')
```

### CSV Files

```python
import csv

# Reading CSV
with open("file.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)  # row is a list

# Reading as dictionary
with open("file.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)  # row is a dict

# Writing CSV
with open("file.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["col1", "col2"])          # Single row
    writer.writerows([[1, 2], [3, 4]])          # Multiple rows

# Writing dictionary
with open("file.csv", "w", newline="") as f:
    fieldnames = ["name", "age"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({"name": "Alice", "age": 25})
```

### JSON Files

```python
import json

# Reading JSON
with open("file.json", "r") as f:
    data = json.load(f)             # Returns dict/list

# Writing JSON
with open("file.json", "w") as f:
    json.dump(data, f, indent=4)    # Pretty print with indent

# String conversion
json_string = json.dumps(data, indent=2)    # Dict to JSON string
data = json.loads(json_string)              # JSON string to dict
```

## 📌 Pathlib (Modern File Operations)

```python
from pathlib import Path

# Creating paths
p = Path("folder/file.txt")
p = Path("folder") / "subfolder" / "file.txt"   # Join paths

# Path properties
p.name          # "file.txt"
p.stem          # "file"
p.suffix        # ".txt"
p.parent        # Path("folder/subfolder")
p.absolute()    # Full path

# File operations
p.exists()          # Check if exists
p.is_file()         # Check if file
p.is_dir()          # Check if directory
p.read_text()       # Read file content
p.write_text("data") # Write to file
p.read_bytes()      # Read binary
p.write_bytes(b"data") # Write binary

# Directory operations
p.mkdir(exist_ok=True)              # Create directory
p.mkdir(parents=True)               # Create with parents
p.rmdir()                           # Remove empty directory
p.unlink()                          # Delete file

# Listing files
p.iterdir()                         # List contents (not recursive)
p.glob("*.txt")                     # Find files by pattern
p.rglob("*.py")                     # Recursive glob

# Special paths
Path.cwd()          # Current working directory
Path.home()         # User home directory
```

## 📌 Object-Oriented Programming (OOP)

```python
# Basic class
class ClassName:
    # Class variable (shared by all instances)
    class_var = "value"
    
    def __init__(self, param1, param2):
        """Constructor - initialize instance"""
        self.param1 = param1            # Instance variable
        self.param2 = param2
    
    def method(self):
        """Instance method"""
        return self.param1
    
    @classmethod
    def class_method(cls):
        """Class method - works with class, not instance"""
        return cls.class_var
    
    @staticmethod
    def static_method():
        """Static method - doesn't access class or instance"""
        return "static"

# Creating objects
obj = ClassName("val1", "val2")
obj.method()

# Inheritance
class ChildClass(ParentClass):
    def __init__(self, param1, param2, param3):
        super().__init__(param1, param2)    # Call parent constructor
        self.param3 = param3
    
    def method(self):
        # Override parent method
        return super().method() + " extended"
```

### Special Methods (Dunder Methods)

```python
__init__(self)          # Constructor
__str__(self)           # String representation (for print)
__repr__(self)          # Official representation
__len__(self)           # len(obj)
__getitem__(self, key)  # obj[key]
__setitem__(self, key, value) # obj[key] = value
__add__(self, other)    # obj1 + obj2
__eq__(self, other)     # obj1 == obj2
```

### Error Handling in ETL

```python
def safe_process(data):
    """Process data with error handling"""
    errors = []
    results = []
    
    for item in data:
        try:
            result = transform(item)
            results.append(result)
        except Exception as e:
            errors.append({"item": item, "error": str(e)})
    
    return results, errors
```

### Working with Nested Data

```python
# Flatten nested dictionary
def flatten_dict(d, parent_key='', sep='_'):
    items = []
    for k, v in d.items():
        new_key = f"{parent_key}{sep}{k}" if parent_key else k
        if isinstance(v, dict):
            items.extend(flatten_dict(v, new_key, sep=sep).items())
        else:
            items.append((new_key, v))
    return dict(items)

# Example
nested = {"user": {"name": "Alice", "age": 25}}
flat = flatten_dict(nested)  # {"user_name": "Alice", "user_age": 25}
```

## 📌 Performance Tips

1. **Use comprehensions** over loops when possible
2. **Use generators** for large datasets: `(x for x in data)` instead of `[x for x in data]`
3. **Use `set` for membership testing** - O(1) vs O(n) for lists
4. **Use `dict.get()` with default** instead of checking keys
5. **Use `with` statement** for file operations (auto-closes)
6. **Avoid repeated string concatenation** - use `join()` instead
7. **Use `enumerate()` instead of** `range(len())`
8. **Cache expensive operations** when possible
