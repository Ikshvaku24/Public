## Part 1: Working with JSON (JavaScript Object Notation)

### What is JSON?
- JSON is a lightweight data format used for data exchange between systems
- All APIs (OpenAI, Claude, web services) communicate using JSON
- In Python, JSON data looks almost identical to dictionaries

### Why JSON for Agentic AI?
- Agent receives instructions in JSON format
- Agent sends responses back as JSON
- Agent state/memory is stored as JSON
- Tool/function calls use JSON for parameters

````python
import json

# Python dictionary (in-memory data structure)
agent_config = {
    "name": "ResearchAgent",
    "model": "gpt-4",
    "temperature": 0.7,
    "tools": ["search", "calculator"],
    "max_iterations": 5
}

print("Python Dictionary:")
print(agent_config)
print(f"Type: {type(agent_config)}")
````

### Converting Dictionary to JSON String (Serialization)
Use `json.dumps()` to convert Python dict → JSON string (for sending over network/APIs)

````python
# Convert dictionary to JSON string
json_string = json.dumps(agent_config)

print("JSON String:")
print(json_string)
print(f"Type: {type(json_string)}")

# Pretty-print JSON with indentation
json_pretty = json.dumps(agent_config, indent=2)
print("\nPretty JSON:")
print(json_pretty)
````

### Converting JSON String to Dictionary (Deserialization)
Use `json.loads()` to convert JSON string → Python dict (for processing API responses)

````python
# Simulating an API response (this is what you receive from OpenAI/Claude APIs)
api_response = '''
{
    "agent_id": "agent_123",
    "status": "completed",
    "result": {
        "answer": "The capital of France is Paris",
        "confidence": 0.98,
        "sources": ["Wikipedia", "Encyclopedia"]
    },
    "tool_calls": 3
}
'''

# Parse JSON string to Python dictionary
response_dict = json.loads(api_response)

print("Parsed Response:")
print(response_dict)
print(f"\nType: {type(response_dict)}")

# Now you can access like a normal dictionary
print(f"\nAgent Status: {response_dict['status']}")
print(f"Answer: {response_dict['result']['answer']}")
print(f"Confidence: {response_dict['result']['confidence']}")
````

### Writing JSON to File
Use `json.dump()` to save agent configuration or state to a file

````python
# Agent memory that needs to be persisted
agent_memory = {
    "conversation_history": [
        {"role": "user", "content": "What is AI?"},
        {"role": "assistant", "content": "AI stands for Artificial Intelligence..."}
    ],
    "context": {"user_name": "Alice", "topic": "AI basics"},
    "session_id": "session_001"
}

# Write to file
with open('agent_memory.json', 'w') as file:
    json.dump(agent_memory, file, indent=2)

print("Agent memory saved to 'agent_memory.json'")
````

### Reading JSON from File
Use `json.load()` to restore agent state from a file

````python
# Read from file
with open('agent_memory.json', 'r') as file:
    loaded_memory = json.load(file)

print("Loaded Agent Memory:")
print(json.dumps(loaded_memory, indent=2))

# Access the conversation history
print("\nConversation History:")
for message in loaded_memory['conversation_history']:
    print(f"{message['role']}: {message['content']}")
````

## Part 2: Object-Oriented Programming (OOP) - Classes and Objects

### What is OOP?
- **Object-Oriented Programming** is a paradigm where we organize code into **objects** that contain data (attributes) and behavior (methods)
- Think of it as creating blueprints (classes) and building things from those blueprints (objects)

### Why OOP for Agentic AI?
- Each **agent is an object** with its own state (memory, configuration)
- **Methods** define what the agent can do (think, search, respond)
- Multiple agents can be created from the same **class** blueprint
- Enables **modular, reusable, and maintainable** agent architectures

### Key Concepts:
- **Class**: A blueprint/template for creating objects
- **Object**: An instance created from a class
- **Attributes**: Variables that belong to an object (state/data)
- **Methods**: Functions that belong to an object (behavior/actions)
- **`self`**: Refers to the current instance of the class

### Creating Your First Class

````python
# Define a simple Agent class
class Agent:
    # This is a class - a blueprint for creating agents
    pass  # Empty class for now

# Create an object (instance) from the class
agent1 = Agent()
agent2 = Agent()

print(f"Agent 1: {agent1}")
print(f"Agent 2: {agent2}")
print(f"Are they the same? {agent1 == agent2}")  # Different objects!
````

### The `__init__()` Method (Constructor)
- Special method called automatically when you create an object
- Used to initialize the object's attributes
- `self` refers to the instance being created

````python
class Agent:
    # Constructor method - runs when object is created
    def __init__(self, name, role):
        # Instance attributes (unique to each object)
        self.name = name          # Agent's name
        self.role = role          # Agent's role (researcher, coder, etc.)
        self.memory = []          # Agent's memory (empty list)
        self.task_count = 0       # Number of tasks completed

# Create agents with different attributes
researcher = Agent("ResearchBot", "researcher")
coder = Agent("CodeAssistant", "programmer")

# Access attributes
print(f"Agent 1: {researcher.name}, Role: {researcher.role}")
print(f"Agent 2: {coder.name}, Role: {coder.role}")
print(f"Researcher memory: {researcher.memory}")
````

### Adding Methods (Behaviors)
Methods are functions defined inside a class that operate on the object's data

````python
class Agent:
    def __init__(self, name, role):
        self.name = name
        self.role = role
        self.memory = []
        self.task_count = 0
    
    # Method to introduce the agent
    def introduce(self):
        return f"Hi, I'm {self.name}, a {self.role} agent."
    
    # Method to add information to memory
    def remember(self, information):
        self.memory.append(information)
        print(f"{self.name} remembered: {information}")
    
    # Method to process a task
    def process_task(self, task):
        self.task_count += 1
        result = f"Task '{task}' processed by {self.name}"
        self.remember(f"Completed task: {task}")  # Store in memory
        return result
    
    # Method to show agent status
    def get_status(self):
        return {
            "name": self.name,
            "role": self.role,
            "tasks_completed": self.task_count,
            "memory_items": len(self.memory)
        }

# Create an agent
agent = Agent("DataAnalyzer", "data scientist")

# Call methods
print(agent.introduce())
print()

# Process tasks
print(agent.process_task("Analyze sales data"))
print(agent.process_task("Generate report"))
print()

# Check status
print("Agent Status:")
print(json.dumps(agent.get_status(), indent=2))
print()

print(f"Full Memory: {agent.memory}")
````

### Understanding `self`
- `self` is a reference to the current object
- Always the first parameter in instance methods
- Used to access attributes and methods of the object

````python
# Create multiple agents
agent1 = Agent("Alice", "researcher")
agent2 = Agent("Bob", "coder")

# Each agent has its own state
agent1.remember("Python is great for AI")
agent2.remember("FastAPI is good for backends")

agent1.process_task("Research LangGraph")
agent2.process_task("Build API endpoint")

print(f"\n{agent1.name}'s memory: {agent1.memory}")
print(f"{agent2.name}'s memory: {agent2.memory}")

# Different task counts
print(f"\n{agent1.name} completed {agent1.task_count} tasks")
print(f"{agent2.name} completed {agent2.task_count} tasks")
````

### Class Attributes vs Instance Attributes

````python
class Agent:
    # Class attribute - shared by ALL instances
    total_agents = 0
    framework = "LangGraph"  # All agents use the same framework
    
    def __init__(self, name, role):
        # Instance attributes - unique to each agent
        self.name = name
        self.role = role
        
        # Increment class attribute when new agent is created
        Agent.total_agents += 1
        self.agent_id = Agent.total_agents
    
    def get_info(self):
        return f"Agent #{self.agent_id}: {self.name} ({self.role}) using {Agent.framework}"

# Create agents
a1 = Agent("Alice", "researcher")
a2 = Agent("Bob", "coder")
a3 = Agent("Carol", "writer")

print(a1.get_info())
print(a2.get_info())
print(a3.get_info())

print(f"\nTotal agents created: {Agent.total_agents}")

# Change class attribute affects all instances
Agent.framework = "CrewAI"
print(f"\nAfter framework change:")
print(a1.get_info())
````

### 🎯 Practice Exercise: Build a Conversational Agent

````python
class ConversationalAgent:
    def __init__(self, name, personality="helpful"):
        self.name = name
        self.personality = personality
        self.conversation_history = []
        self.context = {}  # Store user information
    
    def receive_message(self, user_message):
        """Store user message in history"""
        self.conversation_history.append({
            "role": "user",
            "content": user_message
        })
        return self.generate_response(user_message)
    
    def generate_response(self, message):
        """Generate a simple response based on personality"""
        # Simple rule-based response
        if "name" in message.lower() and "my name is" in message.lower():
            # Extract name (simple extraction)
            parts = message.lower().split("my name is")
            if len(parts) > 1:
                user_name = parts[1].strip().split()[0].capitalize()
                self.context["user_name"] = user_name
                response = f"Nice to meet you, {user_name}! I'm {self.name}."
        elif "hello" in message.lower() or "hi" in message.lower():
            response = f"Hello! I'm {self.name}, how can I help you?"
        else:
            response = f"I understand you said: '{message}'. How can I assist further?"
        
        # Store response
        self.conversation_history.append({
            "role": "assistant",
            "content": response
        })
        return response
    
    def get_conversation_summary(self):
        """Return summary of the conversation"""
        return {
            "agent_name": self.name,
            "message_count": len(self.conversation_history),
            "context": self.context,
            "history": self.conversation_history
        }

# Test the conversational agent
chatbot = ConversationalAgent("Alex", personality="friendly")

print(chatbot.receive_message("Hello!"))
print(chatbot.receive_message("My name is Rahul"))
print(chatbot.receive_message("What can you do?"))

print("\n" + "="*50)
print("Conversation Summary:")
print(json.dumps(chatbot.get_conversation_summary(), indent=2))
````

## Part 3: Error Handling

### Why Error Handling in Agentic AI?
- APIs can fail (network errors, rate limits)
- LLM outputs can be malformed JSON
- Tool calls can have invalid parameters
- Agents should handle errors gracefully and continue operation

### The `try-except` Block
Allows you to catch errors and handle them without crashing

````python
# Example 1: Division by zero
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None

print(safe_divide(10, 2))   # Works fine
print(safe_divide(10, 0))   # Handles error gracefully
````

````python
# Example 2: Parsing invalid JSON (common in agent workflows)
def parse_agent_response(response_string):
    try:
        data = json.loads(response_string)
        return data
    except json.JSONDecodeError as e:
        print(f"Error parsing JSON: {e}")
        return {"error": "Invalid JSON format"}

# Valid JSON
valid_json = '{"status": "success", "result": "Done"}'
print("Valid JSON:", parse_agent_response(valid_json))

# Invalid JSON (missing closing brace)
invalid_json = '{"status": "success", "result": "Done"'
print("Invalid JSON:", parse_agent_response(invalid_json))
````

````python
# Example 3: Handling multiple error types
def access_agent_memory(memory_dict, key):
    try:
        value = memory_dict[key]
        # Try to convert to integer
        return int(value)
    except KeyError:
        print(f"Error: Key '{key}' not found in memory")
        return None
    except ValueError:
        print(f"Error: Cannot convert '{value}' to integer")
        return None
    except Exception as e:
        print(f"Unexpected error: {e}")
        return None

memory = {"task_count": "5", "agent_name": "Alice"}

print(access_agent_memory(memory, "task_count"))    # Works
print(access_agent_memory(memory, "missing_key"))   # KeyError
print(access_agent_memory(memory, "agent_name"))    # ValueError
````

### Using `try-except-else-finally`

````python
def save_agent_state(agent_data, filename):
    try:
        # Try to save the data
        with open(filename, 'w') as f:
            json.dump(agent_data, f, indent=2)
    except IOError as e:
        print(f"Error saving file: {e}")
    else:
        # Executes only if no error occurred
        print(f"Successfully saved agent state to {filename}")
    finally:
        # Always executes (cleanup code)
        print("Save operation completed")

agent_state = {"name": "Agent1", "status": "active"}
save_agent_state(agent_state, "agent_state.json")
````

### 🎯 Practice: Build a Robust Agent with Error Handling

````python
class RobustAgent:
    def __init__(self, name):
        self.name = name
        self.memory = {}
    
    def call_tool(self, tool_name, parameters):
        """Simulate calling an external tool with error handling"""
        try:
            # Simulate tool execution
            if tool_name == "calculator":
                result = parameters["a"] / parameters["b"]
                return {"status": "success", "result": result}
            elif tool_name == "search":
                query = parameters["query"]
                return {"status": "success", "results": [f"Result for {query}"]}
            else:
                raise ValueError(f"Unknown tool: {tool_name}")
        
        except KeyError as e:
            return {"status": "error", "message": f"Missing parameter: {e}"}
        except ZeroDivisionError:
            return {"status": "error", "message": "Division by zero"}
        except Exception as e:
            return {"status": "error", "message": f"Tool error: {str(e)}"}
    
    def load_memory_from_file(self, filename):
        """Load memory from JSON file with error handling"""
        try:
            with open(filename, 'r') as f:
                self.memory = json.load(f)
            print(f"✓ Memory loaded from {filename}")
            return True
        except FileNotFoundError:
            print(f"✗ File {filename} not found. Starting with empty memory.")
            return False
        except json.JSONDecodeError:
            print(f"✗ Invalid JSON in {filename}. Starting with empty memory.")
            return False

# Test the robust agent
agent = RobustAgent("SafeAgent")

print("Test 1: Valid calculator call")
print(agent.call_tool("calculator", {"a": 10, "b": 2}))

print("\nTest 2: Division by zero")
print(agent.call_tool("calculator", {"a": 10, "b": 0}))

print("\nTest 3: Missing parameter")
print(agent.call_tool("calculator", {"a": 10}))

print("\nTest 4: Unknown tool")
print(agent.call_tool("unknown_tool", {}))

print("\nTest 5: Load memory from non-existent file")
agent.load_memory_from_file("nonexistent.json")

print("\nTest 6: Load memory from existing file")
agent.load_memory_from_file("agent_memory.json")
print(f"Loaded memory: {agent.memory}")
````

## 🚀 BONUS Part 4: List Comprehensions

### What are List Comprehensions?
- Concise way to create lists in Python
- Very common in data processing for agents
- More readable and faster than traditional loops

### Syntax:
```python
[expression for item in iterable if condition]
```

````python
# Traditional way
numbers = [1, 2, 3, 4, 5]
squares = []
for num in numbers:
    squares.append(num ** 2)
print("Traditional:", squares)

# List comprehension way
squares = [num ** 2 for num in numbers]
print("Comprehension:", squares)
````

````python
# Example: Filter conversation history
conversation = [
    {"role": "user", "content": "Hello"},
    {"role": "assistant", "content": "Hi there!"},
    {"role": "user", "content": "How are you?"},
    {"role": "assistant", "content": "I'm doing well!"}
]

# Extract only user messages
user_messages = [msg["content"] for msg in conversation if msg["role"] == "user"]
print("User messages:", user_messages)

# Get message lengths
message_lengths = [len(msg["content"]) for msg in conversation]
print("Message lengths:", message_lengths)
````

## 🚀 BONUS Part 5: Working with `*args` and `**kwargs`

### Why for Agentic AI?
- Agents need to accept variable numbers of arguments
- Tool functions have different parameter requirements
- Enables flexible agent architectures

### `*args` - Variable Positional Arguments

````python
def process_multiple_tasks(*tasks):
    """Agent can process any number of tasks"""
    print(f"Processing {len(tasks)} tasks...")
    for i, task in enumerate(tasks, 1):
        print(f"  Task {i}: {task}")
    return f"Completed {len(tasks)} tasks"

# Can call with any number of arguments
print(process_multiple_tasks("Research AI"))
print()
print(process_multiple_tasks("Code review", "Write tests", "Deploy app"))
````

### `**kwargs` - Variable Keyword Arguments

````python
def configure_agent(**config):
    """Agent accepts any configuration parameters"""
    print("Agent Configuration:")
    for key, value in config.items():
        print(f"  {key}: {value}")
    return config

# Can pass any keyword arguments
configure_agent(name="Agent1", model="gpt-4", temperature=0.7)
print()
configure_agent(name="Agent2", role="researcher", max_tokens=1000, verbose=True)
````

````python
# Combining both in a flexible agent class
class FlexibleAgent:
    def __init__(self, name, **config):
        self.name = name
        self.config = config
    
    def execute_tool(self, tool_name, *args, **kwargs):
        """Execute any tool with flexible arguments"""
        print(f"[{self.name}] Executing tool: {tool_name}")
        print(f"  Positional args: {args}")
        print(f"  Keyword args: {kwargs}")
        return {"tool": tool_name, "status": "executed"}

agent = FlexibleAgent("FlexAgent", model="gpt-4", temperature=0.7)

# Can call with different argument patterns
agent.execute_tool("search", "AI agents", max_results=5, filter="recent")
print()
agent.execute_tool("calculator", 10, 20, operation="add")
````

## 🎯 Final Exercise: Build a Complete Agent System

**Challenge**: Create a multi-agent system with error handling and persistence

````python
class AgentSystem:
    def __init__(self):
        self.agents = []
        self.task_history = []
    
    def add_agent(self, agent):
        self.agents.append(agent)
        print(f"✓ Agent '{agent.name}' added to system")
    
    def delegate_task(self, task, agent_name):
        """Delegate task to specific agent with error handling"""
        try:
            # Find agent
            agent = next(a for a in self.agents if a.name == agent_name)
            result = agent.process_task(task)
            
            # Record in history
            self.task_history.append({
                "task": task,
                "agent": agent_name,
                "status": "completed"
            })
            return result
        
        except StopIteration:
            error_msg = f"Agent '{agent_name}' not found"
            self.task_history.append({
                "task": task,
                "agent": agent_name,
                "status": "failed",
                "error": error_msg
            })
            return {"error": error_msg}
    
    def save_system_state(self, filename="system_state.json"):
        """Save entire system state"""
        try:
            state = {
                "agents": [a.get_status() for a in self.agents],
                "task_history": self.task_history
            }
            with open(filename, 'w') as f:
                json.dump(state, f, indent=2)
            print(f"✓ System state saved to {filename}")
        except Exception as e:
            print(f"✗ Error saving state: {e}")
    
    def get_system_report(self):
        """Generate comprehensive system report"""
        total_tasks = len(self.task_history)
        completed = len([t for t in self.task_history if t["status"] == "completed"])
        
        return {
            "total_agents": len(self.agents),
            "agent_names": [a.name for a in self.agents],
            "total_tasks": total_tasks,
            "completed_tasks": completed,
            "failed_tasks": total_tasks - completed
        }

# Test the complete system
system = AgentSystem()

# Create and add agents
researcher = Agent("ResearchBot", "researcher")
coder = Agent("CodeBot", "programmer")
writer = Agent("WriteBot", "writer")

system.add_agent(researcher)
system.add_agent(coder)
system.add_agent(writer)

print("\n" + "="*50)
print("Delegating tasks...\n")

# Delegate tasks
system.delegate_task("Research LangGraph framework", "ResearchBot")
system.delegate_task("Write API documentation", "WriteBot")
system.delegate_task("Implement agent class", "CodeBot")
system.delegate_task("Debug system", "NonExistentBot")  # This will fail

print("\n" + "="*50)
print("System Report:")
print(json.dumps(system.get_system_report(), indent=2))

print("\n" + "="*50)
system.save_system_state()

print("\n✅ Agent system successfully implemented!")
````
