## 1. DataFrames and Series

### Series - 1D Labeled Array
A Series is like a column in a spreadsheet - one-dimensional labeled array that can hold any data type.

```python
# Creating a Series from a list
temperatures = pd.Series([25, 28, 30, 27, 26], name='Temperature')
print("Series from list:")
print(temperatures)
print(f"\nData type: {temperatures.dtype}")

# Creating a Series with custom index
cities = pd.Series(
    [25, 28, 30, 27, 26],
    index=['Delhi', 'Mumbai', 'Bangalore', 'Chennai', 'Kolkata'],
    name='Temperature'
)
print("\nSeries with custom index:")
print(cities)

# Creating a Series from dictionary
population = pd.Series({
    'Delhi': 16.8,
    'Mumbai': 12.5,
    'Bangalore': 8.4,
    'Chennai': 7.1
}, name='Population (Million)')
print("\nSeries from dictionary:")
print(population)
```

```python
# Series operations
print("Series Properties:")
print(f"Shape: {cities.shape}")
print(f"Size: {cities.size}")
print(f"Index: {cities.index.tolist()}")
print(f"Values: {cities.values}")

# Accessing elements
print(f"\nDelhi temperature: {cities['Delhi']}")
print(f"First temperature: {cities.iloc[0]}")

# Statistical operations
print(f"\nMean: {cities.mean():.2f}")
print(f"Max: {cities.max()}")
print(f"Min: {cities.min()}")
print(f"Std Dev: {cities.std():.2f}")
```

### DataFrame - 2D Labeled Data Structure

## 2. Data Loading - Parquet and Excel Files

### Creating Sample Data Files
First, let's create sample files to work with.

```python
# Create sample sales data
sample_data = {
    'Order_ID': range(1001, 1021),
    'Customer_Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'] * 4,
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headset'] * 4,
    'Quantity': np.random.randint(1, 10, 20),
    'Price': [50000, 500, 1500, 15000, 2000] * 4,
    'Order_Date': pd.date_range('2024-01-01', periods=20, freq='D')
}

df_sales = pd.DataFrame(sample_data)
df_sales['Total'] = df_sales['Quantity'] * df_sales['Price']
print("Sample sales data:")
print(df_sales.head(10))
```

### Working with Parquet Files
Parquet is a columnar storage format optimized for analytics - faster and more efficient than CSV.

```python
# Save DataFrame to Parquet
df_sales.to_parquet('sales_data.parquet', index=False)
print("✓ Data saved to Parquet format")

# Read from Parquet
df_from_parquet = pd.read_parquet('sales_data.parquet')
print("\nData loaded from Parquet:")
print(df_from_parquet.head())
print(f"\nShape: {df_from_parquet.shape}")
print(f"Memory usage: {df_from_parquet.memory_usage(deep=True).sum() / 1024:.2f} KB")

# Read specific columns only (column pruning)
df_subset = pd.read_parquet('sales_data.parquet', columns=['Order_ID', 'Customer_Name', 'Total'])
print("\nReading specific columns only:")
print(df_subset.head())
```

### Working with Excel Files

```python
# Save DataFrame to Excel
df_sales.to_excel('sales_data.xlsx', sheet_name='Sales', index=False)
print("✓ Data saved to Excel format")

# Read from Excel
df_from_excel = pd.read_excel('sales_data.xlsx', sheet_name='Sales')
print("\nData loaded from Excel:")
print(df_from_excel.head())

# Read with specific parameters
df_excel_params = pd.read_excel(
    'sales_data.xlsx',
    sheet_name='Sales',
    usecols=['Order_ID', 'Customer_Name', 'Product', 'Total'],  # Select columns
    nrows=10  # Read only first 10 rows
)
print("\nReading with parameters (specific columns, limited rows):")
print(df_excel_params)
```

```python
# Writing multiple sheets to Excel
with pd.ExcelWriter('sales_report.xlsx', engine='openpyxl') as writer:
    df_sales.to_excel(writer, sheet_name='All_Sales', index=False)
    df_sales[df_sales['Product'] == 'Laptop'].to_excel(writer, sheet_name='Laptop_Sales', index=False)
    df_sales.groupby('Product')['Total'].sum().to_excel(writer, sheet_name='Summary')

print("✓ Multi-sheet Excel file created")

# Reading specific sheet
df_laptop = pd.read_excel('sales_report.xlsx', sheet_name='Laptop_Sales')
print("\nLaptop sales sheet:")
print(df_laptop)
```

## 3. Data Cleaning

### Handling Null/Missing Values

```python
# Create DataFrame with missing values
data_with_nulls = {
    'Employee_ID': [1, 2, 3, 4, 5, 6],
    'Name': ['Alice', 'Bob', None, 'Diana', 'Eve', 'Frank'],
    'Age': [28, None, 35, 29, None, 42],
    'Department': ['IT', 'HR', 'IT', None, 'Finance', 'IT'],
    'Salary': [80000, 70000, 90000, 75000, 85000, None]
}

df_nulls = pd.DataFrame(data_with_nulls)
print("DataFrame with null values:")
print(df_nulls)

# Checking for null values
print("\nNull values per column:")
print(df_nulls.isnull().sum())

print("\nPercentage of null values:")
print((df_nulls.isnull().sum() / len(df_nulls) * 100).round(2))
```

```python
# Handling null values - Different strategies

# 1. Drop rows with any null value
df_dropped = df_nulls.dropna()
print("After dropping rows with any null:")
print(df_dropped)

# 2. Drop rows where specific columns have nulls
df_dropped_subset = df_nulls.dropna(subset=['Name', 'Department'])
print("\nAfter dropping rows with null in Name or Department:")
print(df_dropped_subset)

# 3. Fill null values with specific value
df_filled = df_nulls.fillna({'Name': 'Unknown', 'Age': 0, 'Department': 'Unassigned', 'Salary': 0})
print("\nAfter filling nulls with specific values:")
print(df_filled)

# 4. Fill with statistical measures
df_filled_mean = df_nulls.copy()
df_filled_mean['Age'] = df_filled_mean['Age'].fillna(df_filled_mean['Age'].mean())
df_filled_mean['Salary'] = df_filled_mean['Salary'].fillna(df_filled_mean['Salary'].median())
print("\nAfter filling Age with mean and Salary with median:")
print(df_filled_mean)

# 5. Forward fill (use previous value)
df_ffill = df_nulls.fillna(method='ffill')
print("\nAfter forward fill:")
print(df_ffill)
```

### Handling Duplicates

```python
# Create DataFrame with duplicates
data_with_dupes = {
    'Order_ID': [1, 2, 3, 2, 4, 5, 3],
    'Customer': ['Alice', 'Bob', 'Charlie', 'Bob', 'Diana', 'Eve', 'Charlie'],
    'Amount': [100, 200, 300, 200, 400, 500, 300]
}

df_dupes = pd.DataFrame(data_with_dupes)
print("DataFrame with duplicates:")
print(df_dupes)

# Check for duplicates
print("\nDuplicate rows:")
print(df_dupes.duplicated())
print(f"\nTotal duplicates: {df_dupes.duplicated().sum()}")

# Find duplicate rows
print("\nRows that are duplicates:")
print(df_dupes[df_dupes.duplicated(keep=False)])
```

```python
# Remove duplicates

# Keep first occurrence
df_no_dupes_first = df_dupes.drop_duplicates(keep='first')
print("After removing duplicates (keep first):")
print(df_no_dupes_first)

# Keep last occurrence
df_no_dupes_last = df_dupes.drop_duplicates(keep='last')
print("\nAfter removing duplicates (keep last):")
print(df_no_dupes_last)

# Remove all duplicates
df_no_dupes_all = df_dupes.drop_duplicates(keep=False)
print("\nAfter removing all duplicates:")
print(df_no_dupes_all)

# Remove duplicates based on specific columns
df_no_dupes_subset = df_dupes.drop_duplicates(subset=['Order_ID'], keep='first')
print("\nAfter removing duplicates based on Order_ID:")
print(df_no_dupes_subset)
```

### Data Type Conversions

```python
# Create sample DataFrame
data_types = {
    'ID': ['1', '2', '3', '4', '5'],
    'Price': ['100.50', '200.75', '150.25', '300.00', '175.50'],
    'Quantity': ['10', '20', '15', '25', '30'],
    'Is_Active': ['True', 'False', 'True', 'True', 'False'],
    'Date': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05']
}

df_types = pd.DataFrame(data_types)
print("Original data types:")
print(df_types.dtypes)
print("\nOriginal DataFrame:")
print(df_types)
```

```python
# Convert data types
df_converted = df_types.copy()

# String to integer
df_converted['ID'] = df_converted['ID'].astype(int)

# String to float
df_converted['Price'] = df_converted['Price'].astype(float)

# String to integer
df_converted['Quantity'] = pd.to_numeric(df_converted['Quantity'], errors='coerce')  # Safe conversion

# String to boolean
df_converted['Is_Active'] = df_converted['Is_Active'].map({'True': True, 'False': False})

# String to datetime
df_converted['Date'] = pd.to_datetime(df_converted['Date'])

print("Converted data types:")
print(df_converted.dtypes)
print("\nConverted DataFrame:")
print(df_converted)
```

```python
# Handling errors in conversion
data_with_errors = {'Value': ['100', '200', 'invalid', '300', 'N/A']}
df_errors = pd.DataFrame(data_with_errors)

# Using errors='coerce' to handle invalid values
df_errors['Value_Numeric'] = pd.to_numeric(df_errors['Value'], errors='coerce')
print("Handling conversion errors (invalid values become NaN):")
print(df_errors)
```

## 4. Filtering, Sorting, and Selecting Data

### Selecting Columns

```python
# Create sample DataFrame
employees_df = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Age': [28, 35, 30, 42, 29],
    'Salary': [80000, 70000, 90000, 95000, 75000],
    'Experience': [3, 8, 5, 15, 4]
})

print("Original DataFrame:")
print(employees_df)

# Select single column (returns Series)
print("\nSingle column (Series):")
print(employees_df['Name'])

# Select single column (returns DataFrame)
print("\nSingle column (DataFrame):")
print(employees_df[['Name']])

# Select multiple columns
print("\nMultiple columns:")
print(employees_df[['Name', 'Department', 'Salary']])
```

### Filtering Rows

```python
# Filter with single condition
print("Employees in IT department:")
it_employees = employees_df[employees_df['Department'] == 'IT']
print(it_employees)

# Filter with numeric condition
print("\nEmployees with salary > 75000:")
high_earners = employees_df[employees_df['Salary'] > 75000]
print(high_earners)

# Filter with multiple conditions (AND)
print("\nIT employees with salary > 75000:")
it_high_earners = employees_df[(employees_df['Department'] == 'IT') & (employees_df['Salary'] > 75000)]
print(it_high_earners)

# Filter with multiple conditions (OR)
print("\nEmployees in IT or HR:")
it_or_hr = employees_df[(employees_df['Department'] == 'IT') | (employees_df['Department'] == 'HR')]
print(it_or_hr)

# Filter with NOT
print("\nEmployees NOT in IT:")
not_it = employees_df[~(employees_df['Department'] == 'IT')]
print(not_it)
```

```python
# Filter with isin() for multiple values
print("Employees in IT or Finance:")
selected_depts = employees_df[employees_df['Department'].isin(['IT', 'Finance'])]
print(selected_depts)

# Filter with string methods
print("\nEmployees whose name starts with 'A':")
starts_with_a = employees_df[employees_df['Name'].str.startswith('A')]
print(starts_with_a)

# Filter with between()
print("\nEmployees with age between 30 and 40:")
age_range = employees_df[employees_df['Age'].between(30, 40)]
print(age_range)
```

### Selecting with loc and iloc

```python
# iloc - Integer location based indexing
print("First row using iloc:")
print(employees_df.iloc[0])

print("\nFirst 3 rows, first 3 columns:")
print(employees_df.iloc[0:3, 0:3])

print("\nSpecific rows and columns:")
print(employees_df.iloc[[0, 2, 4], [1, 3, 4]])  # Rows 0,2,4 and columns 1,3,4

# loc - Label based indexing
print("\nUsing loc with row index and column names:")
print(employees_df.loc[0:2, ['Name', 'Department', 'Salary']])

print("\nUsing loc with conditions:")
print(employees_df.loc[employees_df['Salary'] > 80000, ['Name', 'Salary']])
```

### Sorting Data

## 5. DateTime Handling

```python
# Create DataFrame with datetime
dates_data = {
    'Transaction_ID': range(1, 11),
    'Date': pd.date_range('2024-01-01', periods=10, freq='D'),
    'Amount': [100, 150, 200, 175, 300, 250, 125, 400, 350, 275]
}

df_dates = pd.DataFrame(dates_data)
print("DataFrame with datetime:")
print(df_dates)
print(f"\nDate column dtype: {df_dates['Date'].dtype}")
```

```python
# Converting strings to datetime
date_strings = {
    'ID': [1, 2, 3, 4],
    'Date_Str': ['2024-01-15', '15-02-2024', '03/20/2024', '2024.04.25']
}

df_date_str = pd.DataFrame(date_strings)
print("Original date strings:")
print(df_date_str)

# Convert with specific format
df_date_str['Date_Converted'] = pd.to_datetime(df_date_str['Date_Str'], format='mixed', dayfirst=False)
print("\nConverted to datetime:")
print(df_date_str)
```

```python
# Extracting datetime components
df_dates['Year'] = df_dates['Date'].dt.year
df_dates['Month'] = df_dates['Date'].dt.month
df_dates['Day'] = df_dates['Date'].dt.day
df_dates['DayOfWeek'] = df_dates['Date'].dt.dayofweek  # Monday=0, Sunday=6
df_dates['DayName'] = df_dates['Date'].dt.day_name()
df_dates['Quarter'] = df_dates['Date'].dt.quarter
df_dates['WeekOfYear'] = df_dates['Date'].dt.isocalendar().week

print("DataFrame with extracted datetime components:")
print(df_dates)
```

```python
# DateTime arithmetic
df_dates['Next_Week'] = df_dates['Date'] + pd.Timedelta(days=7)
df_dates['Previous_Month'] = df_dates['Date'] - pd.DateOffset(months=1)
df_dates['Days_Since_Start'] = (df_dates['Date'] - df_dates['Date'].min()).dt.days

print("DateTime arithmetic:")
print(df_dates[['Date', 'Next_Week', 'Previous_Month', 'Days_Since_Start']])

# Filter by date range
print("\nTransactions in first week of January:")
filtered = df_dates[(df_dates['Date'] >= '2024-01-01') & (df_dates['Date'] <= '2024-01-07')]
print(filtered[['Transaction_ID', 'Date', 'Amount']])
```

## 6. GroupBy Operations and Aggregations

```python
# Create sales dataset
sales_data = {
    'Region': ['North', 'South', 'North', 'East', 'West', 'South', 'North', 'East', 'West', 'South'],
    'Product': ['Laptop', 'Laptop', 'Mouse', 'Laptop', 'Mouse', 'Keyboard', 'Keyboard', 'Mouse', 'Laptop', 'Mouse'],
    'Sales_Person': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie', 'Alice'],
    'Quantity': [5, 3, 50, 4, 45, 30, 25, 40, 6, 55],
    'Revenue': [250000, 150000, 25000, 200000, 22500, 45000, 37500, 20000, 300000, 27500]
}

df_sales_gb = pd.DataFrame(sales_data)
print("Sales Dataset:")
print(df_sales_gb)
```

```python
# Basic GroupBy with single column
print("Total revenue by Region:")
region_revenue = df_sales_gb.groupby('Region')['Revenue'].sum()
print(region_revenue)

print("\nAverage quantity by Product:")
product_avg = df_sales_gb.groupby('Product')['Quantity'].mean()
print(product_avg)

# Multiple aggregations
print("\nMultiple aggregations for Revenue:")
revenue_stats = df_sales_gb.groupby('Region')['Revenue'].agg(['sum', 'mean', 'count', 'min', 'max'])
print(revenue_stats)
```

```python
# GroupBy with multiple columns
print("Revenue by Region and Product:")
region_product = df_sales_gb.groupby(['Region', 'Product'])['Revenue'].sum()
print(region_product)

# Reset index to convert back to DataFrame
print("\nAs DataFrame:")
print(region_product.reset_index())

# Aggregate multiple columns
print("\nAggregate both Quantity and Revenue:")
multi_agg = df_sales_gb.groupby('Region').agg({
    'Quantity': ['sum', 'mean'],
    'Revenue': ['sum', 'mean', 'max']
})
print(multi_agg)
```

```python
# Custom aggregation functions
def revenue_per_unit(x):
    return x.sum() / len(x)

print("Custom aggregation - Revenue per transaction:")
custom_agg = df_sales_gb.groupby('Region')['Revenue'].agg([
    ('total_revenue', 'sum'),
    ('avg_revenue', 'mean'),
    ('revenue_per_transaction', revenue_per_unit)
])
print(custom_agg)

# Using lambda functions
print("\nUsing lambda for range:")
range_agg = df_sales_gb.groupby('Product')['Revenue'].agg(
    Range=lambda x: x.max() - x.min()
)
print(range_agg)
```

```python
# Transform - apply function and return same-sized object
df_sales_gb['Revenue_Mean_By_Region'] = df_sales_gb.groupby('Region')['Revenue'].transform('mean')
df_sales_gb['Revenue_Pct_Of_Region'] = (df_sales_gb['Revenue'] / df_sales_gb['Revenue_Mean_By_Region'] * 100).round(2)

print("DataFrame with transform results:")
print(df_sales_gb[['Region', 'Product', 'Revenue', 'Revenue_Mean_By_Region', 'Revenue_Pct_Of_Region']])
```

## 7. Merging and Joining Datasets

```python
# Create sample DataFrames
employees = pd.DataFrame({
    'Emp_ID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Dept_ID': [10, 20, 10, 30, 20]
})

departments = pd.DataFrame({
    'Dept_ID': [10, 20, 30, 40],
    'Dept_Name': ['IT', 'HR', 'Finance', 'Marketing'],
    'Location': ['Mumbai', 'Delhi', 'Bangalore', 'Chennai']
})

salaries = pd.DataFrame({
    'Emp_ID': [1, 2, 3, 6],
    'Salary': [80000, 75000, 90000, 85000]
})

print("Employees:")
print(employees)
print("\nDepartments:")
print(departments)
print("\nSalaries:")
print(salaries)
```

### Different Types of Merges

```python
# Inner Join (default) - only matching records
print("Inner Join (Employees + Departments):")
inner_merge = pd.merge(employees, departments, on='Dept_ID', how='inner')
print(inner_merge)

# Left Join - all records from left, matching from right
print("\nLeft Join (Employees + Salaries):")
left_merge = pd.merge(employees, salaries, on='Emp_ID', how='left')
print(left_merge)

# Right Join - all records from right, matching from left
print("\nRight Join (Employees + Salaries):")
right_merge = pd.merge(employees, salaries, on='Emp_ID', how='right')
print(right_merge)

# Outer Join - all records from both
print("\nOuter Join (Employees + Salaries):")
outer_merge = pd.merge(employees, salaries, on='Emp_ID', how='outer')
print(outer_merge)
```

```python
# Merge with different column names
projects = pd.DataFrame({
    'Employee_ID': [1, 2, 3, 4],
    'Project': ['Project_A', 'Project_B', 'Project_C', 'Project_A']
})

print("Merge with different column names:")
merged = pd.merge(employees, projects, left_on='Emp_ID', right_on='Employee_ID', how='inner')
print(merged)

# Multiple key merge
print("\nComplete employee information (chain merges):")
complete = employees.merge(departments, on='Dept_ID').merge(salaries, on='Emp_ID', how='left')
print(complete)
```

### Concatenation

```python
# Create DataFrames for concatenation
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Value': [10, 20, 30]
})

df2 = pd.DataFrame({
    'ID': [4, 5, 6],
    'Value': [40, 50, 60]
})

df3 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Category': ['A', 'B', 'C']
})

# Vertical concatenation (stacking rows)
print("Vertical concatenation (rows):")
vertical = pd.concat([df1, df2], ignore_index=True)
print(vertical)

# Horizontal concatenation (side by side)
print("\nHorizontal concatenation (columns):")
horizontal = pd.concat([df1, df3], axis=1)
print(horizontal)
```

## 8. Pivot Tables and Cross-Tabulations

```python
# Create sample data
sales_pivot = pd.DataFrame({
    'Date': pd.date_range('2024-01-01', periods=12, freq='M'),
    'Region': ['North', 'South', 'East', 'West'] * 3,
    'Product': ['Laptop', 'Laptop', 'Mouse', 'Mouse', 'Keyboard', 'Keyboard'] * 2,
    'Sales': [100, 150, 200, 180, 120, 160, 110, 145, 190, 175, 125, 155],
    'Quantity': [10, 15, 200, 180, 120, 160, 11, 14, 195, 170, 122, 158]
})

print("Sales Data:")
print(sales_pivot)
```

```python
# Basic pivot table
print("Pivot: Region vs Product (Sales):")
pivot1 = sales_pivot.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)
print(pivot1)

# Pivot with multiple aggregations
print("\nPivot with multiple aggregations:")
pivot2 = sales_pivot.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc=['sum', 'mean', 'count']
)
print(pivot2)
```

```python
# Pivot with margins (totals)
print("Pivot with row and column totals:")
pivot3 = sales_pivot.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)
print(pivot3)

# Multiple values in pivot
print("\nPivot with multiple values:")
pivot4 = sales_pivot.pivot_table(
    values=['Sales', 'Quantity'],
    index='Region',
    columns='Product',
    aggfunc='sum'
)
print(pivot4)
```

### Cross-Tabulation

```python
# Cross-tabulation (frequency table)
print("Cross-tabulation: Region vs Product (count):")
crosstab1 = pd.crosstab(sales_pivot['Region'], sales_pivot['Product'])
print(crosstab1)

# Cross-tab with values
print("\nCross-tabulation with Sales values:")
crosstab2 = pd.crosstab(
    sales_pivot['Region'],
    sales_pivot['Product'],
    values=sales_pivot['Sales'],
    aggfunc='sum'
)
print(crosstab2)

# Cross-tab with margins and percentages
print("\nCross-tabulation with percentages:")
crosstab3 = pd.crosstab(
    sales_pivot['Region'],
    sales_pivot['Product'],
    normalize='all'  # 'all', 'index', or 'columns'
).round(3)
print(crosstab3)
```

## 9. Multi-Index DataFrames

```python
# Create multi-index DataFrame
arrays = [
    ['North', 'North', 'South', 'South', 'East', 'East'],
    ['2024-Q1', '2024-Q2', '2024-Q1', '2024-Q2', '2024-Q1', '2024-Q2']
]

index = pd.MultiIndex.from_arrays(arrays, names=['Region', 'Quarter'])

df_multi = pd.DataFrame({
    'Sales': [100, 120, 150, 180, 90, 110],
    'Profit': [20, 25, 30, 35, 18, 22]
}, index=index)

print("Multi-Index DataFrame:")
print(df_multi)
print(f"\nIndex levels: {df_multi.index.names}")
```

```python
# Accessing data in multi-index
print("Data for North region:")
print(df_multi.loc['North'])

print("\nData for North, Q1:")
print(df_multi.loc[('North', '2024-Q1')])

# Cross-section
print("\nAll Q1 data across regions:")
print(df_multi.xs('2024-Q1', level='Quarter'))

# Slicing multi-index
print("\nSlicing from North to South:")
print(df_multi.loc['North':'South'])
```

```python
# Manipulating multi-index

# Reset index (convert to columns)
print("Reset index:")
df_reset = df_multi.reset_index()
print(df_reset)

# Set multi-index
print("\nSet multi-index from columns:")
df_set = df_reset.set_index(['Region', 'Quarter'])
print(df_set)

# Swap index levels
print("\nSwap index levels:")
df_swapped = df_multi.swaplevel('Region', 'Quarter')
print(df_swapped)

# Sort by index
print("\nSort by index:")
print(df_multi.sort_index())
```

```python
# Aggregations with multi-index
print("Sum by Region (first level):")
print(df_multi.groupby(level='Region').sum())

print("\nSum by Quarter (second level):")
print(df_multi.groupby(level='Quarter').sum())

# Unstack - pivot inner level to columns
print("\nUnstack (Quarter to columns):")
unstacked = df_multi.unstack(level='Quarter')
print(unstacked)

# Stack - pivot columns to index
print("\nStack back:")
stacked = unstacked.stack()
print(stacked)
```

## Hands-on Exercise: Complex Data Transformation

```python
# Create comprehensive dataset
transactions = pd.DataFrame({
    'Transaction_ID': range(1001, 1021),
    'Customer_ID': [1, 2, 3, 1, 2, 4, 5, 3, 1, 2, 4, 5, 3, 1, 2, 4, 5, 1, 2, 3],
    'Product_ID': [101, 102, 101, 103, 102, 101, 103, 102, 101, 103, 102, 101, 103, 102, 101, 103, 102, 101, 103, 102],
    'Quantity': [2, 1, 3, 1, 2, 1, 2, 1, 3, 1, 2, 1, 3, 2, 1, 2, 1, 3, 2, 1],
    'Amount': [200, 150, 300, 100, 300, 100, 200, 150, 300, 100, 300, 100, 300, 300, 100, 200, 150, 300, 200, 150],
    'Date': pd.date_range('2024-01-01', periods=20, freq='D')
})

customers = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 4, 5],
    'Customer_Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'City': ['Mumbai', 'Delhi', 'Bangalore', 'Chennai', 'Mumbai']
})

products = pd.DataFrame({
    'Product_ID': [101, 102, 103],
    'Product_Name': ['Laptop', 'Mouse', 'Keyboard'],
    'Category': ['Electronics', 'Accessories', 'Accessories']
})

print("Transactions:")
print(transactions.head())
print("\nCustomers:")
print(customers)
print("\nProducts:")
print(products)
```

```python
# Task 1: Merge all data
complete_data = transactions.merge(customers, on='Customer_ID').merge(products, on='Product_ID')
print("Complete merged data:")
print(complete_data.head())

# Task 2: Add time-based features
complete_data['Month'] = complete_data['Date'].dt.month
complete_data['Week'] = complete_data['Date'].dt.isocalendar().week
complete_data['DayOfWeek'] = complete_data['Date'].dt.day_name()

# Task 3: Calculate metrics
complete_data['Revenue'] = complete_data['Quantity'] * complete_data['Amount']

print("\nData with features:")
print(complete_data[['Transaction_ID', 'Customer_Name', 'Product_Name', 'Revenue', 'Month', 'Week']].head())
```

```python
# Task 4: Customer-level aggregations
customer_summary = complete_data.groupby('Customer_Name').agg({
    'Transaction_ID': 'count',
    'Quantity': 'sum',
    'Revenue': ['sum', 'mean', 'max'],
    'Product_Name': lambda x: x.nunique()
}).round(2)

customer_summary.columns = ['Total_Transactions', 'Total_Quantity', 'Total_Revenue', 'Avg_Revenue', 'Max_Revenue', 'Unique_Products']
customer_summary = customer_summary.reset_index()

print("Customer summary:")
print(customer_summary)

# Task 5: Product performance by city
city_product_pivot = complete_data.pivot_table(
    values='Revenue',
    index='City',
    columns='Product_Name',
    aggfunc='sum',
    margins=True
).round(2)

print("\nProduct performance by city:")
print(city_product_pivot)
```

```python
# Task 6: Export results
with pd.ExcelWriter('sales_analysis.xlsx') as writer:
    complete_data.to_excel(writer, sheet_name='Detailed_Data', index=False)
    customer_summary.to_excel(writer, sheet_name='Customer_Summary', index=False)
    city_product_pivot.to_excel(writer, sheet_name='City_Product_Pivot')

print("✓ Complete analysis exported to sales_analysis.xlsx")

# Save summary to CSV
customer_summary.to_csv('customer_summary.csv', index=False)
print("✓ Customer summary exported to CSV")
```
